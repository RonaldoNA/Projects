```julia
using Plots
using SpecialFunctions
using LaTeXStrings
using Measures #Uso para margenes y espacios  
using JLD2 #guardar datos 
using Statistics 
using QuadGK
########## Parameters ---
g = 1.0          
J_val = 1.0
t_fin = 8.0       
j_vals = -19:20   
qt_vals = range(0.0, t_fin, length=33) #CHECK THIS WHEN YOU RUN! THE GRID!!!
######### The coefficients for the superposition state in the central part of the chain, R+1, L-1 with
######### |a⟩ = |...0110...⟩, |b⟩ = |...1010...⟩ and |c⟩ = |...0101...⟩
ρ₊(f) = (1 + sqrt(1-4*f))/2
ρ₋(f) = (1 - sqrt(1-4*f))/2  
fluc = 0.01  #la fluctuacion para los sitios R+1 y L-1 
α₀ = trunc(sqrt(2*ρ₊(fluc) - 1), digits=6, base=10)  
α₁ = trunc(sqrt(1 - ρ₊(fluc)), digits=6, base=10)
α₂ = α₁ 

# The exact density for two non-interacting bosons starting at 0 and 1
function exact_density(j, t, g)
    x = -2.0 * g * t
    return abs2(besselj(-j, x)) + abs2(besselj(-j - 1, x))
end

# Preallocate the matrix
Z = zeros(length(t_vals), length(j_vals))

# Populate the matrix
for (col, j) in enumerate(j_vals)
    for (row, t) in enumerate(t_vals)
        Z[row, col] = exact_density(j, t, g)
    end
end
```

```julia
# --- Generate the Heatmap --- 
Tperiod = t_fin
y_full_ticks = 0:0.25:Tperiod  # Rango de la cuadricula (0 a tfin con pasos de 0.25)
y_labels = vec(range(1, Tperiod + 1))     # Las etiquetas que quiero mostrar especificamente
y_label_indices = findall(x -> x in y_labels, y_full_ticks) 
heatmap(j_vals, t_vals, Z, 
        xlabel =L"i", 
        ylabel =L"t(\tau)", 
        yflip = true,
        title =L"|\dots,1_0, 1_1,\dots\rangle",
        color = cgrad(:oslo, rev=true), #cgrad([:white , :blue])
        size=(600, 700),
        #title=L"Quantum~walk~for~U = %$Uva ~ \theta = %$θ ~g = %$gva",
        colorbar_tickfontsize = font(10),
        # fuentes
        guidefontsize = 25,    # Títulos de ejes (i, t) 
        colorbar_titlefontsize = 24, #Tamaño de la barra de calor
        tickfontsize = 19, top_margin = 11mm, right_margin = 11mm, 
        colorbar=true)
```

```julia
# Expected Current 
J_val = 1.0
# I_{j, j+1} = -2J \sum_m J_{m-j}(2Jt) J_{m-j-1}(2Jt) ⟨n_m(0)⟩
# Our initial state has particles at m = 0 and m = 1.
function expected_current(j, t, J_val)
    x = 2.0 * J_val * t
    # Contribution from particle starting at m = 0
    term_0 = besselj(-j, x) * besselj(-j - 1, x)
    # Contribution from particle starting at m = 1
    term_1 = besselj(1 - j, x) * besselj(-j, x)
    
    return -2.0 * J_val * (term_0 + term_1)
end
# Expected Local Bond Energy
# K_{j, j+1} = 0 everywhere, as derived mathematically
function expected_bond_energy(j, t, J_val)
    return 0.0
end
```

```julia
# Preallocate matrices for heatmaps
Z_current = zeros(length(t_vals), length(j_vals))
Z_energy  = zeros(length(t_vals), length(j_vals))

# Populate the matrices
for (col, j) in enumerate(j_vals)
    for (row, t) in enumerate(t_vals)
        Z_current[row, col] = expected_current(j, t, J_val)
        Z_energy[row, col]  = expected_bond_energy(j, t, J_val)
    end
end
```

```julia
# Generate the Heatmaps 
p1 = heatmap(j_vals, t_vals, Z_current, 
        xlabel =L"j", 
        ylabel =L"t(\tau)", 
        title =L"⟨I_{j, j+1}(t)⟩",
        color = :coolwarm,  
        colorbar_title =L"Flux", 
        guidefontsize = 17,    # Títulos de ejes (i, t) 
        colorbar_titlefontsize = 17,
        tickfontsize = 12)

# We use a standard colormap for energy and lock the limits to show 0
p2 = heatmap(j_vals, t_vals, Z_energy, 
        xlabel = L"j", 
        ylabel =L"t(\tau)", 
        title = L"⟨K_{j, j+1}(t)⟩",
        color = :viridis,
        clims = (-1.0, 1.0), 
        colorbar_title = L"Energy~K")

# Combine plots side-by-side
plot(p1, p2, layout = (1, 2), size = (900, 400), bottom_margin = 5mm, left_margin = 10mm)
```

```julia
# Calculate the initial quantum coherence C
s = 0.1
alpha_0 = sqrt(1 - 2s)
alpha_1 = sqrt(s)
alpha_2 = sqrt(s)
C = sqrt(2) * alpha_0 * (alpha_1 + alpha_2) # Equals 0.8 for s=0.1
```

```julia
# 1. Expected Current (Unchanged by coherence)
function expected_current(j, t, J_val)
    x = 2.0 * J_val * t
    term_0 = besselj(-j, x) * besselj(-j - 1, x)
    term_1 = besselj(1 - j, x) * besselj(-j, x)
    return -2.0 * J_val * (term_0 + term_1)
end

# 2. Expected Local Bond Energy (Driven entirely by coherence C)
function expected_bond_energy(j, t, J_val, C)
    x = 2.0 * J_val * t
    term_squared = besselj(-j, x)^2
    term_cross   = besselj(-j - 1, x) * besselj(1 - j, x)
    return 2.0 * J_val * C * (term_squared - term_cross)
end
```

```julia
# Preallocate matrices
Z_current = zeros(length(t_vals), length(j_vals))
Z_energy  = zeros(length(t_vals), length(j_vals))

# Populate the matrices
for (col, j) in enumerate(j_vals)
    for (row, t) in enumerate(t_vals)
        Z_current[row, col] = expected_current(j, t, J_val)
        Z_energy[row, col]  = expected_bond_energy(j, t, J_val, C)
    end
end
```

```julia
# --- Generate the Heatmaps ---
p1 = heatmap(j_vals, t_vals, Z_current, 
        xlabel = "Bond (j to j+1)", 
        ylabel = "Time (t)", 
        title = "Local Current ⟨I_{j, j+1}(t)⟩",
        color = :coolwarm,  
        colorbar_title = "Flux")

p2 = heatmap(j_vals, t_vals, Z_energy, 
        xlabel = "Bond (j to j+1)", 
        ylabel = "Time (t)", 
        title = "Local Bond Energy ⟨K_{j, j+1}(t)⟩",
        color = :viridis,
        colorbar_title = "Energy")

plot(p1, p2, layout = (1, 2), size = (900, 400))
```

```julia
using SpecialFunctions
using Plots

# --- 1. Initial State Parameters (from your snippet) ---
ρ₊(f) = (1 + sqrt(1 - 4*f)) / 2
ρ₋(f) = (1 - sqrt(1 - 4*f)) / 2

fluc = 0.01  # Fluctuation for |a⟩=|0,1⟩, |b⟩=|-1,1⟩, |c⟩=|0,2⟩ (try 0.05 or 0.1 for stronger asymmetry)
α₀ = trunc(sqrt(2*ρ₊(fluc) - 1), digits=6, base=10)
α₁ = trunc(sqrt(1 - ρ₊(fluc)), digits=6, base=10)
α₂ = α₁

# Free 1D Bessel propagator G_free(n, z) = i^n J_n(z)
function G_free(n::Int, z::Real)
    n_abs = abs(n)
    return (1.0im)^n_abs * besselj(n_abs, z)
end

# --- 2. Exact Relative-Coordinate Bessel + Bound-State Propagator G_K(r, r'; qt) ---
function G_rel_sep(r::Int, rp::Int, z_K::Real, γ::Real; M_cut::Int=20)
    if abs(γ) <= 1.0
        # Pure scattering Bessel series (|γ| <= 1)
        val = G_free(r - rp, z_K) + γ * G_free(r + rp, z_K)
        if abs(1.0 - γ^2) > 1e-14
            pref = 1.0 - γ^2
            γ_pow = 1.0
            for m in 1:M_cut
                val -= pref * γ_pow * G_free(r + rp + 2*m, z_K)
                γ_pow *= γ
            end
        end
        return val
    else
        # Bound state + softened Bessel scattering series (γ > 1, 1/γ < 1)
        γ_inv = 1.0 / γ
        ω_b = 0.5 * z_K * (sqrt(γ) + sqrt(γ_inv))
        bound_part = 0.5 * (γ - γ_inv) * (γ_inv^(0.5 * (r + rp))) * (cis(ω_b) + ((-1)^(r + rp)) * cis(-ω_b))

        scat_part = G_free(r - rp, z_K) + γ_inv * G_free(r + rp, z_K)
        pref = 1.0 - γ_inv^2
        γ_pow = 1.0
        for m in 1:M_cut
            scat_part -= pref * γ_pow * G_free(r + rp - 2*m, z_K)
            γ_pow *= γ_inv
        end
        return bound_part + scat_part
    end
end

function G_rel_doublon(rp::Int, z_K::Real, λ::Real, γ::Real; M_cut::Int=20)
    if abs(λ) < 1e-14
        return 0.0 + 0.0im
    elseif abs(γ) <= 1.0
        val = λ * G_free(rp, z_K)
        if abs(γ - 1.0) > 1e-14
            pref = λ * (γ - 1.0)
            γ_pow = 1.0
            for m in 1:M_cut
                val += pref * γ_pow * G_free(rp + 2*m, z_K)
                γ_pow *= γ
            end
        end
        return val
    else
        γ_inv = 1.0 / γ
        ω_b = 0.5 * z_K * (sqrt(γ) + sqrt(γ_inv))
        bound_part = 0.5 * λ * (1.0 - γ_inv) * (γ_inv^(0.5 * rp)) * (cis(ω_b) + ((-1)^rp) * cis(-ω_b))

        scat_part = λ * γ_inv * G_free(rp, z_K)
        pref = λ * γ_inv * (1.0 - γ_inv)
        γ_pow = 1.0
        for m in 1:M_cut
            scat_part -= pref * γ_pow * G_free(rp - 2*m, z_K)
            γ_pow *= γ_inv
        end
        return bound_part + scat_part
    end
end

# --- 3. Compute Exact Density Profile ⟨n_j(qt)⟩ Across the Lattice ---
function exact_bessel_density(j_vals, qt::Real, θ::Real, α0::Real, α1::Real, α2::Real; N_K::Int=40, R_max::Int=16, M_cut::Int=20)
    # Half-integer momentum grid avoids cos(K/2) == 0 at K = ±π
    K_grid = [-π + 2π * (m - 0.5) / N_K for m in 1:N_K]

    ϕ0_K = zeros(ComplexF64, N_K)
    ϕr_K = zeros(ComplexF64, R_max, N_K)

    for (ik, K) in enumerate(K_grid)
        c_half = cos(0.5 * K)
        z_K = 2.0 * qt * c_half
        λ = sqrt(2.0) * cos(0.5 * (K + θ)) / c_half
        γ = λ^2 - 1.0

        # Initial CoM amplitudes for |ψ(0)⟩ = α0|0,1⟩ + α1|-1,1⟩ + α2|0,2⟩
        ϕ1_init = α0 * cis(-0.5 * K)
        ϕ2_init = α1 + α2 * cis(-K)

        ϕ0_K[ik] = G_rel_doublon(1, z_K, λ, γ; M_cut=M_cut) * ϕ1_init +
                   G_rel_doublon(2, z_K, λ, γ; M_cut=M_cut) * ϕ2_init

        for r in 1:R_max
            ϕr_K[r, ik] = G_rel_sep(r, 1, z_K, γ; M_cut=M_cut) * ϕ1_init +
                          G_rel_sep(r, 2, z_K, γ; M_cut=M_cut) * ϕ2_init
        end
    end

    # Reconstruct real-space density ⟨n_j(qt)⟩ = 2|D_j|^2 + ∑_{r>=1} (|Ψ_{j,j+r}|^2 + |Ψ_{j-r,j}|^2)
    densities = zeros(Float64, length(j_vals))
    for (ij, j) in enumerate(j_vals)
        # Doublon contribution 2|D_j|^2
        D_j = 0.0 + 0.0im
        for (ik, K) in enumerate(K_grid)
            D_j += cis(K * j + 0.5 * θ) * ϕ0_K[ik]
        end
        D_j /= N_K
        n_j = 2.0 * abs2(D_j)

        # Separated two-particle contributions |Ψ_{j, j+r}|^2 + |Ψ_{j-r, j}|^2
        for r in 1:R_max
            Ψ_right = 0.0 + 0.0im
            Ψ_left  = 0.0 + 0.0im
            for (ik, K) in enumerate(K_grid)
                Ψ_right += cis(K * (j + 0.5 * r)) * ϕr_K[r, ik]
                Ψ_left  += cis(K * (j - 0.5 * r)) * ϕr_K[r, ik]
            end
            Ψ_right /= N_K
            Ψ_left  /= N_K
            n_j += abs2(Ψ_right) + abs2(Ψ_left)
        end
        densities[ij] = n_j
    end
    return densities
end
```

```julia
# --- 4. Generate Heatmaps for θ = π, θ = π/3, and Left-Right Asymmetry ---
j_vals  = -11:10
qt_vals = range(0.0, 8.0, length=33) # Scaled time qt = 2Jt

Z_pi  = zeros(length(qt_vals), length(j_vals))
Z_pi3 = zeros(length(qt_vals), length(j_vals))

for (row, qt) in enumerate(qt_vals)
    Z_pi[row, :]  = exact_bessel_density(j_vals, qt, pi,     α₀, α₁, α₂)
    Z_pi3[row, :] = exact_bessel_density(j_vals, qt, π / 2, α₀, α₁, α₂)
end
```

```julia
# Left-Right Asymmetry Δn_j = ⟨n_j(qt)⟩ - ⟨n_{1-j}(qt)⟩ at θ = π/3
Z_asym = Z_pi3 .- Z_pi3[:, end:-1:1]

p1 = heatmap(j_vals, qt_vals, Z_pi, xlabel="Site j", ylabel="Scaled Time qt = 2Jt",
             title="θ = π (Centered High Density + Soft Wings)", color=:viridis)
p2 = heatmap(j_vals, qt_vals, Z_pi3, xlabel="Site j", ylabel="Scaled Time qt = 2Jt",
             title="θ = π/3 (Anyonic Walk)", color=:viridis)
p3 = heatmap(j_vals, qt_vals, Z_asym, xlabel="Site j", ylabel="Scaled Time qt = 2Jt",
             title="Asymmetry ⟨n_j⟩ - ⟨n_{1-j}⟩ at θ = π/3", color=:coolwarm)

plot(p1, p2, p3, layout=(1, 3), size=(1300, 420))
```
